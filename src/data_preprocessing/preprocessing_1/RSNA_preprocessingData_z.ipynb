```python
#  Approach 1
import cv2
import pydicom as dcm
import os
import numpy as np

# GLOBALS
ALLOWED_TAGS=[
    "SOPClassUID",
    "SOPInstanceUID",
    "Modality",
    "PatientID",
    "SliceThickness",
    "StudyInstanceUID",
    "SeriesInstanceUID",
    "InstanceNumber",
    "ImagePositionPatient",
    "ImageOrientationPatient",
    "FrameOfReferenceUID",
    "SamplesPerPixel",
    "PhotometricInterpretation",
    "Rows",
    "Columns",
    "PixelSpacing",
    "BitsAllocated",
    "BitsStored",
    "HighBit",
    "PixelRepresentation",
    "PixelData"
]
BATCH_SIZE=100
# Preprocessing function
import os
import pydicom as dcm
import numpy as np
import cv2
import json

def process_series(series_folder_path, preprocessed_output_path, incomplete_file):
    """
    Preprocess a DICOM series into numpy array + metadata.
    If required metadata is missing, skip and log to incomplete_file.
    """

    # Load all dicom files in series
    dcm_files = [os.path.join(series_folder_path, f) for f in os.listdir(series_folder_path) if f.endswith(".dcm")]
    datasets = []
    for f in dcm_files:
        try:
            datasets.append(dcm.dcmread(f))
        except:
            continue
    # Check metadata availability
    first_ds = datasets[0]
    series_id = getattr(first_ds, "SeriesInstanceUID", "UNKNOWN")

    if len(datasets) == 0:
        print(f"⚠️ No valid dicom files in {series_folder_path}")
        print(f"⚠️ Skipping series {series_id} due to no dicom file")
        with open(incomplete_file, "a") as f:
            f.write(series_id + "\n")
        return



    required_tags = ["ImagePositionPatient", "InstanceNumber","SOPInstanceUID","Modality","SliceThickness"]
    if not all(hasattr(first_ds, tag) for tag in required_tags):
        print(f"⚠️ Skipping series {series_id} due to missing metadata")
        with open(incomplete_file, "a") as f:
            f.write(series_id + "\n")
        return

    # Sort slices
    try:
        datasets.sort(key=lambda ds: float(ds.ImagePositionPatient[2]))
    except Exception:
        datasets.sort(key=lambda ds: int(ds.InstanceNumber))

    # Extract pixel data
    slices = []
    for ds in datasets:
        arr = ds.pixel_array.astype(np.float32)
        arr = cv2.resize(arr, (224, 224))  # resize for ResNet
        arr = (arr - np.min(arr)) / (np.max(arr) - np.min(arr) + 1e-5)  # normalize
        slices.append(arr)

    volume = np.stack(slices, axis=0)  # shape = (num_slices, 224, 224)

    # Save processed volume
    os.makedirs(preprocessed_output_path, exist_ok=True)
    np.save(os.path.join(preprocessed_output_path, f"{series_id}.npy"), volume)

    # Save metadata
    meta = {tag: str(getattr(first_ds, tag, None)) for tag in required_tags}
    with open(os.path.join(preprocessed_output_path, f"{series_id}_meta.json"), "w") as f:
        json.dump(meta, f, indent=2)

    print(f"✅ Processed series {series_id} | slices: {volume.shape[0]}")



# Custom Batch Size preprocessing
for i in range(BATCH_SIZE):

  # get last processed series
  print('Reading progress ....')
  last_entry_file=open(f'{preprocessor_approach_1_dir}/lastentry.txt','r')
  last_entry=last_entry_file.read()
  last_entry_file.close()

  # break if all processed
  if(int(last_entry) >= len(series_folders)): # Use >= to handle cases where last_entry is equal to len
    print('🏆All processed')
    break

  curr_idx=int(last_entry)
  print(f'Processing series {series_folders[curr_idx]}')

  # process
  series_path = os.path.join(org_series_folder_path, series_folders[curr_idx])
  process_series(series_path, series_output_folder_path,f'{preprocessor_approach_1_dir}/incomplete_series.txt')

  # save progress
  print('Saving progress ....')
  last_entry_file=open(f'{preprocessor_approach_1_dir}/lastentry.txt','w')
  last_entry_file.write(str(curr_idx+1))
  last_entry_file.close()

  # Log
  print(f'🚀Processed [{i+1}/{BATCH_SIZE}] series in this batch')


# Overall progress
last_entry_file=open(f'{preprocessor_approach_1_dir}/lastentry.txt','r')
last_entry=last_entry_file.read()
last_entry_file.close()

print(f'Total processed series: {last_entry}')
```

```python
# Resnet embedding generation and plotting to check distortion
import os
import numpy as np
import torch
import torch.nn as nn
import torchvision.models as models
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import pydicom as dcm
import random
import cv2

# --- CONFIG ---
series_id = "1.2.826.0.1.3680043.8.498.11527986509512933171256788651291467752"
preprocessed_output_path = series_output_folder_path

# --- LOAD PREPROCESSED VOLUME ---
volume = np.load(os.path.join(preprocessed_output_path, f"{series_id}.npy"))  # (num_slices,224,224)

# --- PICK RANDOM SLICE ---
rand_idx = random.randint(0, volume.shape[0] - 1)
processed_slice = volume[rand_idx]

# --- LOAD ORIGINAL DICOM SLICE (MATCH BY INSTANCE NUMBER) ---
series_path = os.path.join(org_series_folder_path, series_id)
dcm_files = [os.path.join(series_path, f) for f in os.listdir(series_path) if f.endswith(".dcm")]
datasets = [dcm.dcmread(f) for f in dcm_files]
datasets.sort(key=lambda ds: float(ds.ImagePositionPatient[2]) if hasattr(ds, "ImagePositionPatient") else int(ds.InstanceNumber))

raw_slice = datasets[rand_idx].pixel_array.astype(np.float32)

# Resize raw slice to 224x224 for visual comparison
raw_slice_resized = cv2.resize(raw_slice, (224, 224))

# --- PLOT RAW vs PROCESSED SLICE ---
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(raw_slice_resized, cmap="gray")
plt.title("Raw Slice (Resized)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(processed_slice, cmap="gray")
plt.title("Processed Slice (Normalized)")
plt.axis("off")

plt.show()

# --- PREPARE RESNET MODEL ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
resnet = models.resnet50(pretrained=True)
resnet = nn.Sequential(*list(resnet.children())[:-1])  # remove classifier head
resnet = resnet.to(device).eval()

# --- TRANSFORM ---
transform = transforms.Compose([
    transforms.ToTensor(),  # (H,W)->(C,H,W)
    transforms.Normalize(mean=[0.485], std=[0.229])  # single channel normalization (adapted from ImageNet gray)
])

# expand to 3 channels because ResNet expects RGB
volume_3ch = np.repeat(volume[..., np.newaxis], 3, axis=-1)  # (num_slices,224,224,3)

# convert to torch tensor
input_tensor = torch.stack([transforms.ToTensor()(s) for s in volume_3ch])  # (num_slices,3,224,224)

# apply ImageNet normalization
imagenet_transform = transforms.Normalize(mean=[0.485, 0.456, 0.406],
                                          std=[0.229, 0.224, 0.225])
input_tensor = torch.stack([imagenet_transform(img) for img in input_tensor])

input_tensor = input_tensor.to(device)

# --- FORWARD PASS ---
with torch.no_grad():
    embeddings = resnet(input_tensor)  # shape = (num_slices, 2048, 1, 1)

embeddings = embeddings.squeeze(-1).squeeze(-1)  # (num_slices, 2048)

print(f"✅ Embeddings generated for {embeddings.shape[0]} slices")
print(f"Embedding vector shape per slice: {embeddings.shape[1]}")


```
